```python
import pandas as pd
import numpy as np

df = pd.read_csv("household.csv")

print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
display(df.head())
```

```python
print("Column Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Number of Duplicate Rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
print("Statistical Summary:")
display(df.describe())
```

```python
features = [
    "Family_Size",
    "Monthly_Electricity_kWh",
    "Monthly_Water_Liters",
    "Monthly_LPG_Cylinders",
    "Monthly_Food_Expenditure_INR",
    "Monthly_Transport_Expenditure_INR",
    "Internet_Usage_Hours_Month",
    "Waste_Generation_kg_Month"
]

X = df[features]

print("Selected Features:")
display(X.head())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled_df = pd.DataFrame(
    X_scaled,
    columns=features
)

print("Scaled Data:")
display(X_scaled_df.head())
```

```python
print("Mean of scaled features:")
print(X_scaled_df.mean())

print("\nStandard deviation of scaled features:")
print(X_scaled_df.std())
```

```python
preprocessed_df = X_scaled_df.copy()

preprocessed_df.to_csv(
    "preprocessed_household_data.csv",
    index=False
)

print("Preprocessed dataset saved successfully!")
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt


wcss = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

print("WCSS values:")
for k, value in zip(range(2, 11), wcss):
    print(f"K = {k}: {value:.2f}")
```

```python
plt.figure(figsize=(8, 5))

plt.plot(range(2, 11), wcss, marker='o')

plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS / Inertia")
plt.title("Elbow Method for Optimal K")

plt.xticks(range(2, 11))
plt.grid(True)

plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)

print("Silhouette Scores:")

for k, score in zip(range(2, 11), silhouette_scores):
    print(f"K = {k}: {score:.4f}")
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    silhouette_scores,
    marker='o'
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")

plt.xticks(range(2, 11))
plt.grid(True)

plt.show()
```

```python
from sklearn.cluster import KMeans

optimal_k = 6

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X_scaled)

print("K-Means clustering completed!")
print("Number of clusters:", optimal_k)
```

```python
df["Cluster"] = cluster_labels

display(df.head(10))
```

```python
cluster_names = {
    0: "High Consumption",
    1: "Very Low Consumption",
    2: "Moderate Consumption",
    3: "Low Consumption",
    4: "Extremely High Consumption",
    5: "Very High Consumption"
}

df["Consumption_Pattern"] = df["Cluster"].map(cluster_names)

display(
    df[
        ["Household_ID", "Cluster", "Consumption_Pattern"]
    ].head(20)
)
```

```python


planning_summary = df.groupby("Consumption_Pattern").agg(
    Households=("Household_ID", "count"),
    Avg_Electricity_kWh=("Monthly_Electricity_kWh", "mean"),
    Avg_Water_Liters=("Monthly_Water_Liters", "mean"),
    Avg_LPG_Cylinders=("Monthly_LPG_Cylinders", "mean"),
    Avg_Waste_kg=("Waste_Generation_kg_Month", "mean")
).reset_index()

planning_summary
```

```python
def planning_recommendation(pattern):
    if pattern == "Very Low Consumption":
        return "Basic resource support and infrastructure access"
    elif pattern == "Low Consumption":
        return "Essential resource planning"
    elif pattern == "Moderate Consumption":
        return "Balanced resource allocation"
    elif pattern == "High Consumption":
        return "Demand management and conservation"
    elif pattern == "Very High Consumption":
        return "High-demand resource monitoring"
    elif pattern == "Extremely High Consumption":
        return "Priority demand management and infrastructure planning"

planning_summary["Government_Planning_Action"] = (
    planning_summary["Consumption_Pattern"]
    .apply(planning_recommendation)
)

planning_summary
```
